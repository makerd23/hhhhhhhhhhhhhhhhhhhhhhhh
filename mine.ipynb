```python
a=float(input('輸入身高(公尺)'))
b=float(input('輸入體重(公斤)'))
c=b/(a**2)
if c < 18.5:
    print('體重過輕')
elif 18.5 <= c < 24:
    print('正常範圍')
elif 24 <= c < 27:
    print('過重')
elif 27 <= c < 30:
    print('輕度肥胖')
elif 30 <= c < 35:
    print('中度肥胖')
else:
    print('重度肥胖')
```

```python
x=float(input('輸入使用者x座標: '))
y=float(input('輸入使用者y座標: '))
r=10
d=(x**2+y**2)**0.5
if d <= r:
    print('被炸彈波及到')
else:
    print('沒有被炸彈波及到')
```

```python
import random
print('剪刀石頭布遊戲')
print('1: 剪刀  2: 石頭  3: 布')
player=int(input('請輸入你的選擇(1-3): '))
computer=random.randint(1,3)
if player==1:
    print('你出的是 剪刀')
elif player==2:
    print('你出的是 石頭')
elif player==3:
    print('你出的是 布')
else:
    print('輸入錯誤')
    raise SystemExit
if computer==1:
    print('電腦出的是 剪刀')
elif computer==2:
    print('電腦出的是 石頭')
else:
    print('電腦出的是 布')
if player==computer:
    print('平手')
elif (player==1 and computer==3) or (player==2 and computer==1) or (player==3 and computer==2):
    print('你贏了!')
else:
    print('電腦贏了!')
```
